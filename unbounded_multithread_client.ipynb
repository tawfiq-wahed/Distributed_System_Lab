```python
import socket
import threading
import time
```

```python
HOST="127.0.0.1"
PORT=65002
N=100
```

```python
def worker(i):
    with socket.socket(socket.AF_INET,socket.SOCK_STREAM) as s:
        s.connect((HOST,PORT))
        s.sendall(f"request-{i}".encode());
        reply=s.recv(1024).decode()
        print(f"[client{i} ] is giving a {reply}")
        
```

```python
def main():
    start=time.time()
    threads=[threading.Thread(target=worker,args=(i,)) for i in range(N+1)]
    for t in threads:
        t.start()
    for t in threads:
        t.join()
    elapsed=time.time()-start
    print(f"\nAll {N} requests completed in {elapsed:.2f} seconds")
    print("(With a threaded server this should be ~2s, not ~{}s)".format(N * 2))
      
    
    
```

```python
if __name__=="__main__":
    main()
```
