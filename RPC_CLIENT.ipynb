```python
#RPC SERVER AND CLIENT CREATE AND EXECUTE OPERATIONS
import xmlrpc.client
```

```python
HOST="127.0.0.1"
```

```python
PORT=8000
```

```python
def main():
    proxy = xmlrpc.client.ServerProxy(f"http://{HOST}:{PORT}/")
    print("calling remote add(a,b)->",proxy.add(5,7))
    print("calling remote factorial(n)->",proxy.factorial(5))
    print("calling remote reverse_string('SOFTWARE')->",proxy.reverse_string("SOFTWARE"))
    print("is prime or not prime(5) ", proxy.prime(5))
    
```

```python
if __name__=="__main__":
    main()
```
