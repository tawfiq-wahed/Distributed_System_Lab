```python
#RPC SERVER AND CLIENT CREATE AND EXECUTE OPERATIONS
import xmlrpc.client
```

```python
HOST="127.0.0.1"
PORT=65002
```

```python
def main():
    proxy=xmlrpc.client.ServerProxy(f"http://{HOST}:{PORT}/")
    print(proxy.add(5,7))
    print(proxy.reverse_string("halal"))
```

```python
if __name__=="__main__":
    main()
```
