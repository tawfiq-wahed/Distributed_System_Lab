```python
#client server architecture connection->server
import socket
import threading
import time
```

```python
HOST="127.0.0.1"
PORT=65001
tawfiq=socket.socket(socket.AF_INET,socket.SOCK_STREAM)
tawfiq.bind((HOST,PORT))
tawfiq.listen(5)

```

```python
conn,addr=tawfiq.accept()
request=conn.recv(1024).decode()
print(request)
n=int(request)
result=n*n*n*n
reply=f"{result}"
conn.sendall(reply.encode())
conn.close()


```
