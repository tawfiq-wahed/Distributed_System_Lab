```python
import socket
import threading
import time
```

```python
HOST="127.0.0.1"
PORT=65002
```

```python
def handle_client(conn,addr):
    with conn:
        print(f"[server] {threading.get_ident()} waiting for slot")
        request=conn.recv(1024).decode()
        time.sleep(2)
        reply =(f"server {request} is procceded by {threading.get_ident()}")
        conn.sendall(reply.encode());
        print (f"server {request} is done by {threading.get_ident()}")
        
        
```

```python
def main():
    with socket.socket(socket.AF_INET,socket.SOCK_STREAM) as ssocket:
        ssocket.setsockopt(socket.SOL_SOCKET,socket.SO_REUSEADDR,1)
        ssocket.bind((HOST,PORT))
        ssocket.listen(100)
        print(f"[dispatcher] listening on host and port")
        while True:
            conn,addr=ssocket.accept()
            print(f"[dispatcher] spawning thread {addr}")
            worker=threading.Thread(target=handle_client,args=(conn,addr))
            worker.start()
    
    
```

```python
if __name__=="__main__":
    main()
```
