```python
#RPC SERVER AND CLIENT CREATE AND EXECUTE OPERATIONS
from xmlrpc.server import SimpleXMLRPCServer
```

```python
HOST="127.0.0.1"
PORT=8000
```

```python
def add(a,b):
    print(f"[server] add {a} and {b} remotely")
    return a+b;
```

```python
def factorial(n):
    print(f"[server] print factorial of [n]: ")
    result=1
    for i in range(2,n+1):
        result*=i
        return result;
        
```

```python
def reverse_string(s):
    print(f"[server] is reversing string({s!r}) remotely")
    return s[::-1]   
```

```python
def prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
        
    
```

```python
import threading

def main():
    server = SimpleXMLRPCServer((HOST, PORT), allow_none=True)

    server.register_function(add, "add")
    server.register_function(factorial, "factorial")
    server.register_function(reverse_string, "reverse_string")
    server.register_function(prime, "prime")

    print(f"RPC SERVER listening on {HOST}:{PORT}")
    print("[server] exposed procedures: add, factorial, reverse_string, prime")

    server_thread = threading.Thread(
        target=server.serve_forever,
        daemon=True
    )

    server_thread.start()

    return server
```

```python
if __name__=="__main__":
    main()
```
